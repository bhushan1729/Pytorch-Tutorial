```python
import torch.nn as nn

class MitchellVisualCortex(nn.Module):
  def __init__(self):
    super().__init__()
    self.conv1 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.conv2 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.conv3 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.conv4 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.relu = nn.ReLU()
    self.fc1 = nn.Linear(3*120*120, 2)

  def forward(self, x):
    x = self.conv1(x)
    x = self.relu(x)
    x = self.conv2(x)
    x = self.relu(x)
    x = self.conv3(x)
    x = self.relu(x)
    x = self.conv4(x)
    x = self.relu(x)
    # --- THE FLATTENING BRIDGE ---
    # x.size(0) automatically grabs the active batch size dimension dynamically.
    # -1 commands PyTorch to unroll all remaining dimensions (3*120*120) into a continuous row.
    x = x.view(x.size(0), -1)
    x = self.fc1(x)


    return x
```

```python
import torch

# 1. Instantiate a live instance of your custom book network
model = MitchellVisualCortex()

# 2. Simulate 1 fake color image with size 128x128
# Shape format: [Batch_size, Channels, Height, Width]
fake_image = torch.randn(1, 3, 128, 128)

# 3. Test execution!
try:
    output_scores = model(fake_image)
    print("\n🎉 Architecture Sanity Check: SUCCESS!")
    print(f"Input Data Grid Shape  : {fake_image.shape}")
    print(f"Output Class Score Shape: {output_scores.shape}") # Should be [1, 2]
except Exception as e:
    print(f"\n❌ Structural Mismatch Error: {e}")
```

```python
from torchvision import transforms

# 1. Your clean transformation blueprint
img_transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

# 2. This helper function takes a batch dictionary from Hugging Face
# and returns the transformed images along with their labels
def transform_batch(batch):
    # Apply your resizing and tensor conversion to every single PIL image in the incoming batch
    batch["image"] = [img_transform(img) for img in batch["image"]]
    return batch

# 3. Inject the transform into the datasets using the Hugging Face engine
# This changes how data is outputted dynamically without consuming extra RAM!
transformed_train_set = dataset['train'].with_transform(transform_batch)
transformed_test_set = dataset['test'].with_transform(transform_batch)
```

```python
from torch.utils.data import DataLoader

# 4. Instantiate the trucks passing the dynamically transformed wrappers
train_loader = DataLoader(transformed_train_set, batch_size=32, shuffle=True)
test_loader = DataLoader(transformed_test_set, batch_size=32, shuffle=False) # Good practice: don't shuffle testing evaluation metrics
```

```python
# Grab a single real delivery bundle from your training line
first_batch = next(iter(train_loader))

print("--- Pipeline Verification Success ---")
print(f"Batch Images Shape: {first_batch['image'].shape}")   # Expected: [32, 3, 128, 128]
print(f"Batch Labels Shape: {first_batch['labels'].shape}")  # Expected: [32]
```

```python
import torch
import torch.optim as optim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f'using {device}')

model = MitchellVisualCortex().to(device)

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
for epoch in range(10):
  running_loss = 0

  for idx, batch in enumerate(train_loader):
    image = batch['image'].to(device)
    label = batch['labels'].to(device)

    image = image.to(device)
    label = label.to(device)

    prediction = model(image)

    loss = criterion(prediction, label)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    running_loss += loss.item()

  epoch_loss = running_loss / len(train_loader)
  print(f"Epoch {epoch+1} Loss: {epoch_loss}")
```

```python
def calculate_accuracy(data_loader, dataset_name="Dataset"):
    # 1. Flip the model's internal state switch to evaluation mode
    model.eval()

    correct_predictions = 0
    total_samples = 0

    # 2. Freeze the autograd history engine to maximize cluster speed
    with torch.no_grad():
        for batch in data_loader:
            # Unpack the Hugging Face dictionary and send tensors to the active hardware
            images = batch["image"].to(device)
            labels = batch["labels"].to(device)

            # Forward pass to get unnormalized logit scores
            outputs = model(images)

            # Find the index of the highest class value along dimension 1
            _, predicted_classes = torch.max(outputs, dim=1)

            # Accumulate performance metrics math
            total_samples += labels.size(0)
            correct_predictions += (predicted_classes == labels).sum().item()

    accuracy = (correct_predictions / total_samples) * 100
    print(f"🎯 {dataset_name} Accuracy: {accuracy:.2f}% ({correct_predictions}/{total_samples})")
    return accuracy

# --- RUN EXECUTION ---
print("=== Evaluating Mitchell Visual Cortex Performance ===")
train_acc = calculate_accuracy(train_loader, "Training Set")
test_acc  = calculate_accuracy(test_loader, "Test Set")
```

```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# 1. Grab a single image entry out of your test warehouse conveyor belt
model.eval() # Set to evaluation mode
single_entry = next(iter(train_loader))

# Extract the very first single image from the batch
# DataLoader returns a batch dimension [32, 3, 128, 128], we slice out index 0 -> [1, 3, 128, 128]
input_image = single_entry["image"][10].unsqueeze(0).to(device)
true_label = single_entry["labels"][10].item()

# 2. Sequential Extractions: Run data manually through each layer to trap the states
with torch.no_grad():
    # Pass through layer 1 cortex
    act1 = model.relu(model.conv1(input_image))   # Shape: [1, 3, 126, 126]

    # Pass through layer 2 cortex
    act2 = model.relu(model.conv2(act1))          # Shape: [1, 3, 124, 124]

    # Pass through layer 3 cortex
    act3 = model.relu(model.conv3(act2))          # Shape: [1, 3, 122, 122]

    # Pass through layer 4 cortex
    act4 = model.relu(model.conv4(act3))          # Shape: [1, 3, 120, 120]

# List containing our activation states and their text labels for the canvas loop
activation_layers = [act1, act2, act3, act4]
layer_names = ["Conv Layer 1", "Conv Layer 2", "Conv Layer 3", "Conv Layer 4"]

# 3. SETUP MATPLOTLIB PLOTTING CANVAS METRICS
# 4 Rows (Layers) x 3 Columns (Filters)
fig, axes = plt.subplots(4, 3, figsize=(10, 12))

print("=== INTERNAL ACTIVATION MAP NUMERICAL MATRICES (TRUNCATED LOOKUP) ===\n")

for row_idx, layer_tensor in enumerate(activation_layers):
    # Squeeze away the batch dimension: from [1, 3, H, W] down to [3, H, W]
    layer_maps = layer_tensor.squeeze(0).cpu().numpy()

    print(f"--- {layer_names[row_idx]} Spatial Resolution Matrix: {layer_maps.shape[1]}x{layer_maps.shape[2]} ---")

    for col_idx in range(3):
        # Extract the specific 2D activation matrix slice for this filter channel
        filter_matrix = layer_maps[col_idx]

        # PRINT NUMERICAL VALUE CHECK: Print out a tiny 4x4 coordinate section of raw activation levels
        print(f"  > Filter {col_idx+1} Top-Left 4x4 Values:")
        print(np.round(filter_matrix[:4, :4], 4))

        # PLOT VISUALIZATION: Plot the grid matrix on our subplots layout
        ax = axes[row_idx, col_idx]

        # Using 'gray' colormap to show intensity: higher values = white (high activation firing rate)
        ax.imshow(filter_matrix, cmap='gray')
        ax.set_title(f"{layer_names[row_idx]} | Filter {col_idx+1}")
        ax.axis('off')
    print("\n" + "="*60 + "\n")

plt.suptitle(f"Mitchell Cortex Internal Firing Maps | True Label Target Class: {true_label}", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
