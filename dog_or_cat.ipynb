```python
import torch.nn as nn

class MitchellVisualCortex(nn.Module):
  def __init__(self):
    super().__init__()
    self.conv1 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.conv2 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.conv3 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.conv4 = nn.Conv2d(in_channels=3, out_channels=3, kernel_size=3)
    self.relu = nn.ReLU()
    self.fc1 = nn.Linear(3*120*120, 2)

  def forward(self, x):
    x = self.conv1(x)
    x = self.relu(x)
    x = self.conv2(x)
    x = self.relu(x)
    x = self.conv3(x)
    x = self.relu(x)
    x = self.conv4(x)
    x = self.relu(x)
    # --- THE FLATTENING BRIDGE ---
    # x.size(0) automatically grabs the active batch size dimension dynamically.
    # -1 commands PyTorch to unroll all remaining dimensions (3*120*120) into a continuous row.
    x = x.view(x.size(0), -1)
    x = self.fc1(x)


    return x
```

```python
import torch

# 1. Instantiate a live instance of your custom book network
model = MitchellVisualCortex()

# 2. Simulate 1 fake color image with size 128x128
# Shape format: [Batch_size, Channels, Height, Width]
fake_image = torch.randn(1, 3, 128, 128)

# 3. Test execution!
try:
    output_scores = model(fake_image)
    print("\n🎉 Architecture Sanity Check: SUCCESS!")
    print(f"Input Data Grid Shape  : {fake_image.shape}")
    print(f"Output Class Score Shape: {output_scores.shape}") # Should be [1, 2]
except Exception as e:
    print(f"\n❌ Structural Mismatch Error: {e}")
```

```python
from torchvision import transforms

# 1. Your clean transformation blueprint
img_transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

# 2. This helper function takes a batch dictionary from Hugging Face
# and returns the transformed images along with their labels
def transform_batch(batch):
    # Apply your resizing and tensor conversion to every single PIL image in the incoming batch
    batch["image"] = [img_transform(img) for img in batch["image"]]
    return batch

# 3. Inject the transform into the datasets using the Hugging Face engine
# This changes how data is outputted dynamically without consuming extra RAM!
transformed_train_set = dataset['train'].with_transform(transform_batch)
transformed_test_set = dataset['test'].with_transform(transform_batch)
```

```python
from torch.utils.data import DataLoader

# 4. Instantiate the trucks passing the dynamically transformed wrappers
train_loader = DataLoader(transformed_train_set, batch_size=32, shuffle=True)
test_loader = DataLoader(transformed_test_set, batch_size=32, shuffle=False) # Good practice: don't shuffle testing evaluation metrics
```

```python
import torch
import torch.optim as optim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f'using {device}')

model = MitchellVisualCortex().to(device)

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
for epoch in range(10):
  running_loss = 0

  for idx, batch in enumerate(train_loader):
    image = batch['image'].to(device)
    label = batch['labels'].to(device)

    image = image.to(device)
    label = label.to(device)

    prediction = model(image)

    loss = criterion(prediction, label)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    running_loss += loss.item()

  epoch_loss = running_loss / len(train_loader)
  print(f"Epoch {epoch+1} Loss: {epoch_loss}")
```

```python
def calculate_accuracy(data_loader, dataset_name="Dataset"):
    # 1. Flip the model's internal state switch to evaluation mode
    model.eval()

    correct_predictions = 0
    total_samples = 0

    # 2. Freeze the autograd history engine to maximize cluster speed
    with torch.no_grad():
        for batch in data_loader:
            # Unpack the Hugging Face dictionary and send tensors to the active hardware
            images = batch["image"].to(device)
            labels = batch["labels"].to(device)

            # Forward pass to get unnormalized logit scores
            outputs = model(images)

            # Find the index of the highest class value along dimension 1
            _, predicted_classes = torch.max(outputs, dim=1)

            # Accumulate performance metrics math
            total_samples += labels.size(0)
            correct_predictions += (predicted_classes == labels).sum().item()

    accuracy = (correct_predictions / total_samples) * 100
    print(f"🎯 {dataset_name} Accuracy: {accuracy:.2f}% ({correct_predictions}/{total_samples})")
    return accuracy

# --- RUN EXECUTION ---
print("=== Evaluating Mitchell Visual Cortex Performance ===")
train_acc = calculate_accuracy(train_loader, "Training Set")
test_acc  = calculate_accuracy(test_loader, "Test Set")
```

```python
import torch.nn as nn

class MitchellVisualCortex(nn.Module):
  def __init__(self):
    super().__init__()
    self.conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3)
    self.conv2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3)
    self.conv3 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3)
    self.conv4 = nn.Conv2d(in_channels=64, out_channels=128, kernel_size=3)
    self.relu = nn.ReLU()
    self.fc1 = nn.Linear(in_features=128 * 6 * 6, out_features=2)
    self.maxpool = nn.MaxPool2d(kernel_size=2, stride=2)
  def forward(self, x):
    x = self.conv1(x)
    x = self.relu(x)
    x = self.maxpool(x)

    x = self.conv2(x)
    x = self.relu(x)
    x = self.maxpool(x)

    x = self.conv3(x)
    x = self.relu(x)
    x = self.maxpool(x)

    x = self.conv4(x)
    x = self.relu(x)
    x = self.maxpool(x)
    # --- THE FLATTENING BRIDGE ---
    # x.size(0) automatically grabs the active batch size dimension dynamically.
    # -1 commands PyTorch to unroll all remaining dimensions (3*120*120) into a continuous row.
    x = x.view(x.size(0), -1)
    x = self.fc1(x)


    return x

import torch
import torch.optim as optim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f'using {device}')

model = MitchellVisualCortex().to(device)

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(model.parameters(), lr=0.001)


for epoch in range(20):
  running_loss = 0

  for idx, batch in enumerate(train_loader):
    image = batch['image'].to(device)
    label = batch['labels'].to(device)

    image = image.to(device)
    label = label.to(device)

    prediction = model(image)

    loss = criterion(prediction, label)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    running_loss += loss.item()

  epoch_loss = running_loss / len(train_loader)
  print(f"Epoch {epoch+1} Loss: {epoch_loss}")
```

```python
def calculate_accuracy(data_loader, dataset_name="Dataset"):
    # 1. Flip the model's internal state switch to evaluation mode
    model.eval()

    correct_predictions = 0
    total_samples = 0

    # 2. Freeze the autograd history engine to maximize cluster speed
    with torch.no_grad():
        for batch in data_loader:
            # Unpack the Hugging Face dictionary and send tensors to the active hardware
            images = batch["image"].to(device)
            labels = batch["labels"].to(device)

            # Forward pass to get unnormalized logit scores
            outputs = model(images)

            # Find the index of the highest class value along dimension 1
            _, predicted_classes = torch.max(outputs, dim=1)

            # Accumulate performance metrics math
            total_samples += labels.size(0)
            correct_predictions += (predicted_classes == labels).sum().item()

    accuracy = (correct_predictions / total_samples) * 100
    print(f"🎯 {dataset_name} Accuracy: {accuracy:.2f}% ({correct_predictions}/{total_samples})")
    return accuracy

# --- RUN EXECUTION ---
print("=== Evaluating Mitchell Visual Cortex Performance ===")
train_acc = calculate_accuracy(train_loader, "Training Set")
test_acc  = calculate_accuracy(test_loader, "Test Set")
```

```python
import torch
import matplotlib.pyplot as plt
import math

# 1. Prepare a single validation image from your test truck conveyor belt
model.eval()
single_image_batch = next(iter(test_loader))["image"][2].unsqueeze(0).to(device)

# Assemble your newly evolved architecture layers into an ordered tracking list
# NOTE: We include your maxpool operations since your forward pass now uses them!
layers = [model.conv1, model.conv2, model.conv3, model.conv4]
current_features = single_image_batch

# 2. Sequential Extraction Pass
with torch.no_grad():
    for layer_idx, conv_layer in enumerate(layers):

        # Mirror your model's forward pass mechanics: Conv -> ReLU -> MaxPool
        current_features = model.maxpool(model.relu(conv_layer(current_features)))

        # Squeeze out the batch dimension to get a clean 3D array: [Channels, H, W]
        activation_grid = current_features.squeeze(0).cpu().numpy()
        num_filters = activation_grid.shape[0]

        # --- DYNAMIC GRID panel OVERHEAD MATH ---
        # Calculate how many rows and columns we need to form a near-square grid layout panel
        cols = int(math.ceil(math.sqrt(num_filters)))
        rows = int(math.ceil(num_filters / cols))

        # Open a new global figure canvas pane for this entire layer sheet panel
        fig, axes = plt.subplots(rows, cols, figsize=(cols * 1.5, rows * 1.5))
        # Flatten the axes array matrix so we can loop through it cleanly in a single row index list
        axes = axes.flatten() if num_filters > 1 else [axes]

        print(f"📊 Layer {layer_idx + 1} | Unrolling {num_filters} Filters | Map Resolution: {activation_grid.shape[1]}x{activation_grid.shape[2]}")

        # 3. Inner Loop: Render every single filter map stamp into its matrix slot
        for i in range(len(axes)):
            ax = axes[i]
            if i < num_filters:
                # Isolate the exact 2D feature map channel
                target_map = activation_grid[i]
                ax.imshow(target_map, cmap='gray')

            # Keep the layout clean by stripping away axis margins and numbers
            ax.axis('off')

        plt.suptitle(f"Mitchell Visual Cortex — Layer {layer_idx + 1} Full Activation Sheet Panel ({num_filters} Filters)", fontsize=14, fontweight='bold', y=0.98)
        plt.tight_layout()
        plt.show()
        print("\n" + "="*80 + "\n")
```

```python
import torch
import matplotlib.pyplot as plt

# --- PRE-LOOP TRACKERS ---
train_losses = []
test_losses = []

print(f"🚀 Starting Loss Optimization Analysis on {device}...")

for epoch in range(10):
    # ==========================
    # PHASE A: TRAINING ENGINE
    # ==========================
    model.train() # Tell PyTorch we are actively optimizing parameters
    running_train_loss = 0.0

    for batch in train_loader:
        image = batch['image'].to(device)
        label = batch['labels'].to(device)

        # Optimization pass steps
        prediction = model(image)
        loss = criterion(prediction, label)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_train_loss += loss.item()

    avg_train_loss = running_train_loss / len(train_loader)
    train_losses.append(avg_train_loss)

    # ==========================
    # PHASE B: VALIDATION ENGINE
    # ==========================
    model.eval() # Tell PyTorch to freeze parameter tracking behaviors
    running_test_loss = 0.0

    with torch.no_grad(): # Disable the autograd history calculation graph
        for batch in test_loader:
            test_image = batch['image'].to(device)
            test_label = batch['labels'].to(device)

            test_prediction = model(test_image)
            test_loss = criterion(test_prediction, test_label)

            running_test_loss += test_loss.item()

    avg_test_loss = running_test_loss / len(test_loader)
    test_losses.append(avg_test_loss)

    print(f"Epoch {epoch+1:02d}/10 | Train Loss: {avg_train_loss:.4f} | Test Loss: {avg_test_loss:.4f}")

print("\n🎉 Optimization complete! Metrics successfully stored.")
```

```python
plt.figure(figsize=(10, 6))

# Plot both optimization lines
plt.plot(range(1, 11), train_losses, label='Training Loss (Bias Tracker)', color='#1f77b4', marker='o', linewidth=2)
plt.plot(range(1, 11), test_losses, label='Test Loss (Variance Tracker)', color='#ff7f0e', marker='s', linewidth=2)

# Chart layout styling
plt.title("Mitchell Visual Cortex: Loss Optimization Landscape Curve", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Epochs", fontsize=12)
plt.ylabel("Cross Entropy Loss Value", fontsize=12)
plt.xticks(range(1, 11))
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=12, loc='upper right')

# Tight margins formatting
plt.tight_layout()
plt.show()
```
